# RDF Data Loading from Fuseki

This notebook demonstrates how to load RDF data from Apache Jena Fuseki into Spark DataFrames for Big Data Analytics.


```python
# Import required libraries
import sys
import os

# Add scripts directory to path
sys.path.append('/home/jovyan/work/scripts')

from pyspark.sql import SparkSession
from rdf_sparkml import RdfDataLoader
import pandas as pd
```

```python
# Initialize Spark Session
# Connect to Spark cluster
spark = SparkSession.builder \
    .appName("RDF_Data_Loading") \
    .master("spark://spark-master:7077") \
    .config("spark.driver.memory", "2g") \
    .config("spark.executor.memory", "2g") \
    .config("spark.executor.cores", "2") \
    .config("spark.sql.warehouse.dir", "/tmp/spark-warehouse") \
    .getOrCreate()

print(f"Spark version: {spark.version}")
print(f"Spark master: {spark.sparkContext.master}")
```

```python
# Initialize RDF Data Loader
# Uses environment variables or defaults to localhost
loader = RdfDataLoader(
    spark_session=spark,
    sparql_endpoint="http://fuseki:3030/cruise_kg/sparql",
    username="admin",
    password="admin"
)
```

```python
# Load ports data
ports_df = loader.load_ports_dataframe()

# Display schema
ports_df.printSchema()

# Show first 20 rows
ports_df.show(20, truncate=False)
```
