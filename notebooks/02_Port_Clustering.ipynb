# Port Clustering with K-Means

This notebook demonstrates how to cluster ports using K-Means clustering based on geographic location and capacity.


```python
import sys
sys.path.append('/home/jovyan/work/scripts')

from pyspark.sql import SparkSession
from rdf_sparkml import RdfDataLoader, RdfSparkMlService
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Initialize Spark
spark = SparkSession.builder \
    .appName("Port_Clustering") \
    .master("spark://spark-master:7077") \
    .config("spark.driver.memory", "2g") \
    .config("spark.executor.memory", "2g") \
    .getOrCreate()
```

```python
# Load data
loader = RdfDataLoader(spark)
ports_df = loader.load_ports_dataframe()

# Initialize ML service
ml_service = RdfSparkMlService(spark)
```

```python
# Cluster ports into 5 groups
kmeans_model, clustered_df = ml_service.cluster_ports(
    ports_df,
    num_clusters=5,
    features=['latitude', 'longitude', 'berths_capacity']
)

# Show cluster assignments
clustered_df.select(
    "port_name", "city", "country",
    "latitude", "longitude", "berths_capacity", "cluster"
).show(50, truncate=False)
```
